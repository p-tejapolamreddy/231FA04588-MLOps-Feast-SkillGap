```python
%%writefile feature_store/feature_definitions.py

from datetime import timedelta

from feast import Entity, FeatureView, Field
from feast.types import Float32, Int64
from feast.infra.offline_stores.file_source import FileSource


# --------------------------------------------------
# ENTITY
# --------------------------------------------------

skill = Entity(
    name="skill",
    join_keys=["skill_id"],
    description="Technical skill used for curriculum-industry skill gap analysis"
)


# --------------------------------------------------
# DATA SOURCE
# --------------------------------------------------

skill_gap_source = FileSource(
    name="skill_gap_source",
    path="../data/skill_gap_features.parquet",
    timestamp_field="event_timestamp",
    created_timestamp_column="created_timestamp"
)


# --------------------------------------------------
# FEATURE VIEW
# --------------------------------------------------

skill_gap_features = FeatureView(
    name="skill_gap_features",
    entities=[skill],

    ttl=timedelta(days=365),

    schema=[
        Field(
            name="curriculum_score",
            dtype=Float32
        ),

        Field(
            name="industry_score",
            dtype=Float32
        ),

        Field(
            name="skill_gap",
            dtype=Float32
        ),

        Field(
            name="training_priority",
            dtype=Int64
        )
    ],

    online=True,

    source=skill_gap_source,

    tags={
        "project": "curriculum-industry-skill-gap",
        "course": "MLOps"
    }
)
```

```python
from feast import FeatureStore

store = FeatureStore(
    repo_path="/content/feature_store"
)

print("Feast FeatureStore loaded successfully")
```

```python
feature_df = pd.read_parquet(
    "/content/data/skill_gap_features.parquet"
)

print("Rows:", len(feature_df))
feature_df.head()
```

```python
entity_df = feature_df[
    [
        "skill_id",
        "event_timestamp"
    ]
].copy()

entity_df.head()
```

```python
historical_features = store.get_historical_features(
    entity_df=entity_df,

    features=[
        "skill_gap_features:curriculum_score",
        "skill_gap_features:industry_score",
        "skill_gap_features:skill_gap",
        "skill_gap_features:training_priority"
    ]
).to_df()

print("Historical feature retrieval successful!")
print("Rows:", len(historical_features))

historical_features.head(20)
```

```python
historical_features.to_csv(
    "/content/outputs/historical_features.csv",
    index=False
)

print("Saved:")
print("/content/outputs/historical_features.csv")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

print("ML libraries imported successfully")
```

```python
model_data = historical_features[
    [
        "curriculum_score",
        "industry_score",
        "skill_gap",
        "training_priority"
    ]
].dropna()

X = model_data[
    [
        "curriculum_score",
        "industry_score",
        "skill_gap"
    ]
]

y = model_data["training_priority"]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(
    X_train,
    y_train
)

print("Random Forest model trained successfully!")
```

```python
y_pred = model.predict(X_test)

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("Model Accuracy:", accuracy)
print(
    "Model Accuracy:",
    round(accuracy * 100, 2),
    "%"
)
```

```python
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
import joblib

joblib.dump(
    model,
    "/content/outputs/skill_gap_model.joblib"
)

print("Model saved successfully!")
```
